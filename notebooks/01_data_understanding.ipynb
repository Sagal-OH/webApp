```python
# # Import pandas for data manipulation and os to verify file path
import pandas as pd
import os

file_path = "../data/raw/online_retail_II.xlsx"

print("Working directory:", os.getcwd())   # confirm script location

```

```python
# Load both sheets 
sheet_1 = "Year 2009-2010"
sheet_2 = "Year 2010-2011"

# Read each yearly sheet separately (dataset is split across two tabs)
df_0910 = pd.read_excel(file_path, sheet_name=sheet_1)
df_1011 =  pd.read_excel(file_path, sheet_name=sheet_2)

# checks shape 
print ("2009–2010 shape:", df_0910.shape)
print("2010–2011 shape:", df_1011.shape)
```

```python
# Combine both datasets into a single dataframe and resets index for clean sequencing
df_raw = pd.concat([df_0910, df_1011], ignore_index=True)

print("Combined dataset shape:", df_raw.shape)
```

```python
# Display first 5 rows to inspect structure and values
display(df_raw.head())
```

```python
# # List columns and their dtypes .Checking for type mismatches early 
print(df_raw.columns.tolist())
print (df_raw.dtypes)

print("\nDataset Info:")
df_raw.info()
```

```python
# Compute missing values (count and percentage) per column
print("\nMissing Values (Count):")
display (df_raw.isnull().sum())

print("\nMissing Values (%):")
display ((df_raw.isnull().mean() * 100).round(2))
```

```python
# Count unique values per column to understand data variety 
for col in df_raw.columns:
    print (f"{col}: {df_raw[col].nunique()}")
```

```python
# Count duplicate rows 
duplicate_count = df_raw.duplicated().sum()
print (f"\nDuplicate Rows: {duplicate_count}")
```

```python
# Generates  summary statistics for numeric columns
display(df_raw.describe())
```

```python
# Identify the minimum and maximum dates in the dataset
print("\nDate Range:")

print("Earliest:", df_raw['InvoiceDate'].min())
print("Latest:  ", df_raw['InvoiceDate'].max())
```
