```python
# Load dataset 
import pandas as pd

file_path = "../data/raw/online_retail_II.xlsx"

df_0910 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
df_1011 = pd.read_excel(file_path, sheet_name="Year 2010-2011")

# Reload both sheets and combines them 
df_raw = pd.concat([df_0910, df_1011], ignore_index=True)

# Total no of rows
total_rows = len(df_raw)

print("Total Rows:", total_rows)
```

```python
 # Count rows where Customer ID is blank
missing_customer =  df_raw['Customer ID'].isnull().sum()

print ("Missing Customer ID:")
print ("Count:", missing_customer)

# Express as percentage of total rows
print ("Percentage:", round((missing_customer / total_rows) * 100, 2))
```

```python
 # Count rows where Description is blank
missing_desc = df_raw['Description'].isnull().sum()

print("Missing Description:")

print("Count:", missing_desc)

# Express as percentage of total rows
print("Percentage:", round((missing_desc / total_rows) * 100, 2))
```

```python
# Invoices starting with 'C' are cancellations.Casting to string first to avoid type errors
cancelled = df_raw['Invoice'].astype(str).str.startswith('C').sum()

print("Cancellation Invoices:")
print( "Count:", cancelled)

# Express as percentage of total rows
print("Percentage:", round((cancelled / total_rows) * 100, 2))
```

```python
# # Count rows where quantity is negative (could indicates returns or reversals)
negative_qty = (df_raw['Quantity'] < 0).sum()

print("Negative Quantity:")
print ("Count:", negative_qty)

# Express as  a percentage of total rows
print("Percentage:", round((negative_qty / total_rows) * 100, 2))
```

```python
 # Count rows where price is exactly zero  (likely samples, errors or gifts)
zero_price = (df_raw['Price'] == 0).sum()

# Count rows where price is negative 
negative_price =  (df_raw['Price'] < 0).sum()

print("Zero Price:")
print ("Count:", zero_price)

#Express as a  percentage 
print("Percentage:", round((zero_price / total_rows) * 100, 2))

print("\nNegative Price:")
print ("Count:", negative_price)

# express as a percentage
print("Percentage:", round((negative_price / total_rows) * 100, 2))
```

```python
# Count duplicate rows
duplicates = df_raw.duplicated().sum()

print ("Duplicate Rows:")
print("Count:", duplicates)

# Express as percentage 
print("Percentage:", round((duplicates / total_rows) * 100, 2))
```
