```python

import pandas as pd

# Load cleaned CSV from disk
df =  pd.read_csv("../data/processed/cleaned_data.csv", parse_dates=['InvoiceDate'])
# parse_dates converts InvoiceDate from string to datetime automatically

print("Loaded dataset shape:", df.shape)

# visually inspect first 5 rows
df.head()
```

```python
#  Store total row count to use as reference throughout validation checks
total_rows = len(df)

print("Total rows:", total_rows)
```

```python
# Count missing values across every cell in the entire dataframe
missing_total  = df.isnull().sum().sum()

print("Test 1 - Missing Values:", "PASS" if missing_total == 0 else "FAIL")

print("Total missing values:", missing_total)
```

```python
# Check for any rows that are completely identical across all columns
duplicates  = df.duplicated().sum()

print("Test 2 - Duplicates:", "PASS" if duplicates == 0 else "FAIL")

print("Duplicate rows:", duplicates)
```

```python
#  Count rows where Quantity is zero or negative (should be none after cleaning) 
invalid_qty = (df['Quantity'] <= 0).sum()

print("Test 3 - Quantity Valid:", "PASS" if invalid_qty == 0 else "FAIL")

print("Invalid quantity rows:", invalid_qty)
```

```python
# Count rows where Price is zero or negative (should be none after cleaning) 
invalid_price = (df['Price'] <= 0).sum()

print("Test 4 — Price Valid:", "PASS" if invalid_price == 0 else "FAIL")

print ("Invalid price rows:", invalid_price)
```

```python
# Check if any invoices starting with 'C' remain 
cancelled_remaining  = df['Invoice'].astype(str).str.startswith('C').sum()

print("Test 5 — No Cancellations:", "PASS" if cancelled_remaining == 0 else "FAIL")

print("Remaining cancellation rows:", cancelled_remaining)
```

```python

# Recalculate Revenue from scratch and compare to stored Revenue column
# Round both to 2dp
mismatches = ((df['Quantity'] * df['Price']).round(2) != df['Revenue'].round(2)).sum()

print("Test 6 — Revenue:", "PASS" if mismatches == 0 else "FAIL", "| Mismatches:", mismatches)
```

```python
# Confirm date range matches original raw dataset (no dates should have been lost)
print("Test 7 - Date Range:")

print("Earliest:", df['InvoiceDate'].min())

print("Latest:  ", df['InvoiceDate'].max())
```

```python
# # Count distinct Customer IDs(confirms customer data is intact after cleaning)
unique_customers = df['Customer ID'].nunique()

print("Test 8 - Unique Customers:", unique_customers)
```
